## Q1. List operations

```python
L = [10, 20, 30, 40, 50, 60, 70, 80]
print("Original L           :", L)

# i) add 200 and 300
L.append(200)
L.append(300)                    # or in one step: L.extend([200, 300])
print("i)   after adding 200, 300:", L)

# ii) remove 10 and 30
L.remove(10)
L.remove(30)
print("ii)  after removing 10, 30:", L)

# iii) ascending order
L.sort()
print("iii) ascending            :", L)

# iv) descending order
L.sort(reverse=True)
print("iv)  descending           :", L)
```

## Q2. Tuple of scores

```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

# i) highest score and its index
highest = max(scores)
print("i)   highest score:", highest, "| index:", scores.index(highest))

# ii) lowest score and how many times it appears
lowest = min(scores)
print("ii)  lowest score :", lowest, "| appears", scores.count(lowest), "times")

# iii) reverse the tuple and return it as a list
reversed_list = list(reversed(scores))          # same as list(scores[::-1])
print("iii) reversed as list:", reversed_list)

# iv) check a score entered by the user (default 76)
target = read_input("iv)  Enter a score to search for: ", 76)
if target in scores:
    print(f"     {target} is present; first occurrence at index {scores.index(target)}")
else:
    print(f"     {target} is not present in the tuple")
```

## Q3. 100 random numbers between 100 and 900

```python
random.seed(1024160138)                         # fixed seed -> same numbers every run (remove this line for fresh numbers)
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:", numbers)


def is_prime(n):
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2
    return True


odd = [n for n in numbers if n % 2 != 0]
even = [n for n in numbers if n % 2 == 0]
prime = [n for n in numbers if is_prime(n)]

print(f"\ni)   Odd numbers   ({len(odd)}):", odd)
print(f"\nii)  Even numbers  ({len(even)}):", even)
print(f"\niii) Prime numbers ({len(prime)}):", prime)
print("\nCheck: odd + even =", len(odd) + len(even), "(must equal 100)")
```

## Q4. Sets of team scores

```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

print("i)   Union (unique scores of both teams)   :", A.union(B))
print("ii)  Intersection (common scores)          :", A.intersection(B))
print("iii) Symmetric difference (exclusive ones) :", A.symmetric_difference(B))
print("iv)  A is a subset of B   :", A.issubset(B))
print("     B is a superset of A :", B.issuperset(A))

# v) remove a score X (entered by the user) from A if it exists
def remove_score(s, x):
    if x in s:
        s.remove(x)
        print(f"     {x} removed. A is now {s}")
    else:
        print(f"     {x} is not present in A")

X = read_input("v)   Enter the score X to remove from A: ", 56, cast=int)
remove_score(A, X)

# demonstration of the 'not present' branch
print("     (demo) trying a score that is not in A:")
remove_score(A, 999)
```

## Q5. Rename the key `city` to `location`

```python
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

sample_dict["location"] = sample_dict.pop("city")   # pop() returns the value and deletes the old key
print(sample_dict)
# expected: {'name': 'Kelly', 'age': 25, 'salary': 8000, 'location': 'New york'}
```

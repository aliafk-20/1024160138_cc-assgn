## Q4. Read a CSV file and show the first five rows (Iris dataset)

The assignment uses the Kaggle file `Iris.csv` (https://www.kaggle.com/datasets/uciml/iris). Kaggle needs a login, so
download it yourself and upload it next to this notebook (in Colab: Files panel → Upload). The cell below uses that
file if it exists. **If it does not exist**, it writes an `Iris.csv` with the same columns and the same 150 flowers
(taken from scikit-learn's copy of the Iris data) so everything still runs, and the code that follows reads it with
`pd.read_csv` exactly as it would read the Kaggle file.

```python
CSV_PATH = "Iris.csv"
if not os.path.exists(CSV_PATH):
    from sklearn.datasets import load_iris
    iris = load_iris()
    names = {0: "Iris-setosa", 1: "Iris-versicolor", 2: "Iris-virginica"}
    tmp = pd.DataFrame(iris.data, columns=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"])
    tmp.insert(0, "Id", range(1, len(tmp) + 1))
    tmp["Species"] = [names[t] for t in iris.target]
    tmp.to_csv(CSV_PATH, index=False)
    print("Iris.csv not found -> created a stand-in with the same layout from scikit-learn's Iris data.")
else:
    print("Using your Iris.csv")

iris_df = pd.read_csv(CSV_PATH)
print("shape:", iris_df.shape)
print(iris_df.head())                      # first five rows
```

## Q6. Employees dataset

```python
employees = pd.DataFrame({
    "Employee_ID":         [101, 102, 103, 104, 105],
    "Name":                ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department":          ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age":                 [29, 34, 41, 28, 38],
    "Salary":              [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date":        ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender":              ["Female", "Male", "Male", "Female", "Male"],
    "Bonus":               [5000, 7000, 6000, 4500, 5000],
    "Rating":              [4.5, 4.0, 3.8, 4.7, 3.5],
})
employees.to_csv("employees.csv", index=False)         # create employees.csv as the question asks

emp = pd.read_csv("employees.csv", parse_dates=["Joining_Date"])
print(emp)
```

```python
# a) shape
print("a) Shape (rows, columns):", emp.shape)

# b) summary: data types and non-null counts
print("\nb) Info:")
emp.info()

# c) descriptive statistics
print("\nc) Descriptive statistics:")
print(emp.describe())

# d) first 5 rows and last 3 rows
print("\nd) First 5 rows:")
print(emp.head(5))
print("\n   Last 3 rows:")
print(emp.tail(3))
```

```python
# e) statistics
print("e) i)   Average salary          :", emp["Salary"].mean())
print("   ii)  Total bonus paid        :", emp["Bonus"].sum())
print("   iii) Youngest employee's age :", emp["Age"].min())
print("   iv)  Highest rating          :", emp["Rating"].max())

# f) sort by Salary, descending
print("\nf) Sorted by Salary (descending):")
print(emp.sort_values("Salary", ascending=False))

# g) performance category
def categorize(r):
    if r >= 4.5:
        return "Excellent"
    elif r >= 4.0:
        return "Good"
    return "Average"

emp["Performance"] = emp["Rating"].apply(categorize)
print("\ng) With performance category:")
print(emp[["Name", "Rating", "Performance"]])

# h) missing values
print("\nh) Missing values per column:")
print(emp.isnull().sum())
print("Any missing values?", bool(emp.isnull().values.any()))

# i) rename Employee_ID -> ID
emp = emp.rename(columns={"Employee_ID": "ID"})
print("\ni) Columns after rename:", list(emp.columns))
```

```python
# j) filters
print("j) i)  More than 5 years of experience:")
print(emp[emp["Years_of_Experience"] > 5])

print("\n   ii) Belonging to the IT department:")
print(emp[emp["Department"] == "IT"])

print("\n   Both conditions together (IT and > 5 years):")
print(emp[(emp["Department"] == "IT") & (emp["Years_of_Experience"] > 5)])

# k) Tax column = 10% of salary (and the salary left after deducting it)
emp["Tax"] = emp["Salary"] * 0.10
emp["Salary_After_Tax"] = emp["Salary"] - emp["Tax"]
print("\nk) Tax added:")
print(emp[["ID", "Name", "Salary", "Tax", "Salary_After_Tax"]])

# l) save the modified DataFrame
emp.to_csv("employees_modified.csv", index=False)
print("\nl) Saved -> employees_modified.csv")
print(pd.read_csv("employees_modified.csv"))
```

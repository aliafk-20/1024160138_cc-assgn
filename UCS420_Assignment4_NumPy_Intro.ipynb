## Q2. Basic array questions

```python
# a) reverse
arr = np.array([1, 2, 3, 6, 4, 5])
print("a) reversed:", arr[::-1])          # also: np.flip(arr)
```

```python
# b) most frequent value(s) and their indices (ties handled: every value with the max count is reported)
def most_frequent(a):
    values, counts = np.unique(a, return_counts=True)
    top = counts.max()
    for v in values[counts == top]:
        print(f"   value {v} occurs {top} times at indices {np.where(a == v)[0]}")

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("b-i)  x:"); most_frequent(x)
print("b-ii) y (1 and 3 tie):"); most_frequent(y)
```

## Q4. `linspace` array, its attributes, and transpose

```python
yourname = np.linspace(10, 100, 25)
print(yourname)
print("dimensions (ndim):", yourname.ndim)
print("shape            :", yourname.shape)
print("total elements   :", yourname.size)
print("dtype            :", yourname.dtype)
print("total bytes      :", yourname.nbytes, "(= size x itemsize =", yourname.size, "x", yourname.itemsize, ")")

# transpose using reshape(): turn the 1-D array into a column (25 x 1)
col = yourname.reshape(25, 1)
print("\nreshape(25, 1) shape:", col.shape)
print("reshape(1, 25) shape:", yourname.reshape(1, 25).shape, "-> its .T is", yourname.reshape(1, 25).T.shape)

# using .T directly on the 1-D array
print("\nyourname.T shape:", yourname.T.shape, "| unchanged?", np.array_equal(yourname.T, yourname))
```

**Can we do the same with `.T`?** Not on a 1-D array. A transpose reverses the order of the axes; a 1-D array has
only one axis, so `yourname.T` returns the same `(25,)` array. To get a column vector you must add an axis first:
`reshape(25, 1)` (or `yourname[:, None]`), or first make it 2-D — `yourname.reshape(1, 25).T` — and then `.T` works.

## Q5. 2-D array `ucs420_<name>`: statistics, reshape, resize

```python
ucs420_yourname = np.array([[10, 20, 30, 40],
                            [50, 60, 70, 80],
                            [90, 15, 20, 35]])
print("array (3x4):\n", ucs420_yourname)
print("mean   :", ucs420_yourname.mean())
print("median :", np.median(ucs420_yourname))
print("max    :", ucs420_yourname.max())
print("min    :", ucs420_yourname.min())
print("unique :", np.unique(ucs420_yourname), "(20 appears twice, so 11 unique values out of 12)")

# reshape -> 4 rows x 3 columns (same 12 elements, new layout)
reshaped_ucs420_yourname = ucs420_yourname.reshape(4, 3)
print("\nreshaped_ucs420_yourname (4x3):\n", reshaped_ucs420_yourname)

# resize -> 2 rows x 3 columns (6 elements: np.resize keeps the first 6 values in order, rest are dropped)
resized_ucs420_yourname = np.resize(ucs420_yourname, (2, 3))
print("\nresized_ucs420_yourname (2x3):\n", resized_ucs420_yourname)
```

**reshape vs resize:** `reshape` only rearranges the same number of elements (12 → 4x3) and returns a view when
possible; `resize` can change the total size — shrinking drops trailing elements (as here), growing repeats the data.
